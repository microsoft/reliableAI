```python
import sys
sys.path.append("../../../../")
import pickle
import csv
import os
import numpy as np
from src.base.pdag import MixedGraph
from sklearn.preprocessing import StandardScaler
```

```python
exp_name = 'exp_3'
observation_len = 5000
intervention_len_list = [2500, 2500, 2500, 1000, 1000, 1000, 1000, 1000, 1000, 500, 500, 500, 250, 250, 250]
intervention_size_list = [2, 2, 2, 5, 5, 5, 5, 5, 5, 10, 10, 10, 20, 20, 20]

for idx, benchmark_name in enumerate(['survey', 'asia', 'sachs',  'child', 'insurance', 'water', 'mildew', 'alarm', 'barley', 'hailfinder', 'hepar2', 'win95pts', 'pathfinder', 'munin1', 'andes']):
    print(f"{'-'*60} {benchmark_name} {'-'*60} \n")
    os.makedirs(f'/home/v-yangzha/iml4c/baselines/dcdi/data/bnlearn/{exp_name}/{benchmark_name}', exist_ok=True)

    intervention_len = intervention_len_list[idx]
    intervention_size = intervention_size_list[idx]

    aug_graph_path = f'/home/v-yangzha/iml4c/datasets/experiment/original/{exp_name}/aug_graphs/{benchmark_name}_aug_graph.txt'
    aug_dataset_path = f'/home/v-yangzha/iml4c/datasets/experiment/original/{exp_name}/aug_samples/{benchmark_name}_aug_dataset.npy'
    int_targets_path = f'/home/v-yangzha/iml4c/datasets/experiment/original/{exp_name}/int_targets/{benchmark_name}_int_targets.pkl'

    aug_graph = MixedGraph(graph_path = aug_graph_path)
    sys_size = len(aug_graph.Nodes)-intervention_size
    aug_DAG = aug_graph.getAdjacencyMatrix()[:sys_size,:sys_size]
    aug_CPDAG = aug_graph.getCPDAG().getAdjacencyMatrix()[:sys_size,:sys_size]

    with open(aug_dataset_path, 'rb') as f:
        aug_data = np.load(f)
    with open(int_targets_path, 'rb') as f:
        targets_list = pickle.load(f)

    
    np.save(f'/home/v-yangzha/iml4c/baselines/dcdi/data/bnlearn/{exp_name}/{benchmark_name}/DAG1.npy', aug_DAG)
    np.save(f'/home/v-yangzha/iml4c/baselines/dcdi/data/bnlearn/{exp_name}/{benchmark_name}/CPDAG1.npy', aug_CPDAG)
    obs_data = aug_data[:observation_len, :sys_size]
    scaler = StandardScaler()
    obs_data = scaler.fit_transform(obs_data)

    int_data = aug_data[:, :sys_size]
    scaler = StandardScaler()
    int_data = scaler.fit_transform(int_data)

    np.save(f'/home/v-yangzha/iml4c/baselines/dcdi/data/bnlearn/{exp_name}/{benchmark_name}/data1.npy', obs_data)
    np.save(f'/home/v-yangzha/iml4c/baselines/dcdi/data/bnlearn/{exp_name}/{benchmark_name}/data_interv1.npy', int_data)

    mask_intervention = []
    regimes = []
    for i, targets in enumerate(targets_list):
        total_len = observation_len if i==0 else intervention_len
        mask_intervention.extend([np.array(targets, dtype=np.int32) for _ in  range(total_len)])
        regimes.extend([i for _ in  range(total_len)])
    
    with open(f'/home/v-yangzha/iml4c/baselines/dcdi/data/bnlearn/{exp_name}/{benchmark_name}/intervention1.csv', 'w', newline="") as f:
        writer = csv.writer(f)
        writer.writerows(mask_intervention)
    
    with open(f'/home/v-yangzha/iml4c/baselines/dcdi/data/bnlearn/{exp_name}/{benchmark_name}/regime1.csv', 'w', newline="") as f:
        writer = csv.writer(f)
        for regime in regimes:
            writer.writerow([regime])
```
